## Final Wait-Time Dataset Decision

Four additional datasets were evaluated for improving SmartCare's wait-time prediction.

The experiments showed that a large number of records does not guarantee a useful model. Some datasets lacked real queue predictors, some contained privacy or leakage risks, and the simulated ER model failed when evaluated after an operational intervention.

Therefore, none of the newly evaluated models will replace the existing SmartCare prototype model.

SmartCare will continue to use an explainable hybrid estimate combining:

- historical department and triage baseline,
- patients ahead,
- active doctors,
- average consultation duration,
- current doctor delay,
- occupancy and urgency adjustments.

The result will remain an indicative waiting-time range, not a guaranteed clinical appointment time.

No raw dataset, personal information, rejected model, or unreviewed experiment will be deployed.
